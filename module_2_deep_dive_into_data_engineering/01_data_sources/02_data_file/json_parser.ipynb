2. Multiple Level

```python
# HINTS
# 1. Use pandas json_normalize to read data from record_path
# 2. Export to CSV as path: 'destination/json'
import pandas as pd
import json
multiple_level_json_file_path = "multiple_level.json"
with open (file=multiple_level_json_file_path, mode="r") as f:
    data = json.load(f)
print(data)
```

```python
#Hardcode
df = pd.json_normalize(data=data['students'])
df['class'] = data['class']
df['school_name'] = data['school_name']
df
```

```python
#option 2
record_path='students'
meta=['school_name', 'class']
df = pd.json_normalize(data=data, record_path=record_path, meta=meta, record_prefix='students_')
df
```

3. Nested Levels

```python
# HINTS
# Approach 1: use record_path + meta
# 1. Open the load JSON into dictionary 
# 2. Use pandas json_normalize to read data from record_path
# 3. Export to CSV as path: 'destination/json'
nested_level_json_file_path = "nested_level.json"

with open(file=nested_level_json_file_path, mode='r') as f:
    data= json.load(f)
data
```

```python
record_path='students'
meta=['school_name', ['class', 'level'], ['class','room'], ['class', 'teacher']]
df = pd.json_normalize(data=data, record_path=record_path, meta=meta)
df
```

```python
# HINTS
# Approach 2: Read seperately and cross join
# 1. Open the load JSON into dictionary 
# 2. Use pandas json_normalize to read data from record_path
# 3. Export to CSV as path: 'destination/json'
```

```python
df1 = pd.json_normalize(data=data)
df1
```

```python
df2 = pd.json_normalize(data=data['students'])
df2
```

```python
df = pd.merge(left=df1, right=df2, how='cross')
df = df.drop(columns='students')
df
```
